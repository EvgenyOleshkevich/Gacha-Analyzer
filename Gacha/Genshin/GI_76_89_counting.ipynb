## WEAPON ZZZ

```python
base_prob = 0.01
sum_prob = 0.02
Math_Exp = 1 / sum_prob  # раз 50
count_incr = 66
full_guarantee = 80
```

```python
def F(x):
    p = base_prob
    q = 1 - p
    if x > 1:
        x = 1
    if x < p:
        x = p

    acc_q = 1
    sum = 0
    for i in range(1, count_incr):
        sum += i * acc_q * p
        acc_q *= q
        
    p = x
    q = 1 - x
    
    for i in range(count_incr, full_guarantee):
        sum += i * acc_q * p
        acc_q *= q
        
    sum += full_guarantee * acc_q
    return sum
```

```python
def Search(x, expected, Func):
    if x > 1 :
        x = 0.999
    if x < base_prob:
        x = base_prob
    
    h = 0.00001
    f1 = 1
    while f1 > 0.0000001:
        f1 = (expected - Func(x)) ** 2
        f2 = (expected - Func(x + h)) ** 2
        x += (f1 - f2) / (h * 1000) 
        if x > 1:
            x = 0.999
        print(x, Func(x))
    
```
